# 在 Jupyter 中启动第四章看板

依次运行下面两个代码单元格。不需要打开或粘贴 app.py、data_model.py、charts.py。
第一个单元格检查当前Jupyter内核的依赖，第二个启动Streamlit服务；点击输出链接查看看板。
图表显示在独立浏览器页面，Jupyter负责启动服务。


```python
import sys
from pathlib import Path
import importlib.util

folder = Path("D:/CP/python可视化课程/《Excel数据可视化 - 从图表到数据大屏》-清华-郭宏远/第四章可视化/Streamlit版")
missing = [name for name in ["streamlit", "plotly", "openpyxl"] if importlib.util.find_spec(name) is None]
print("当前Jupyter内核：", sys.executable)
if missing:
    raise RuntimeError("当前内核缺少依赖，请在新单元格运行 %pip install streamlit plotly openpyxl，然后重新运行")
assert (folder / "app.py").is_file(), "请修改folder为Streamlit版文件夹的实际位置"
print("代码位置：", folder)
print("依赖检查通过")
```

```python
import subprocess, os, socket, time, urllib.request
from IPython.display import display, Markdown

if "process" not in globals() or process.poll() is not None:
    with socket.socket() as sock:
        sock.bind(("127.0.0.1", 0))
        port = sock.getsockname()[1]
    log_path = Path.cwd() / "Streamlit启动日志.txt"
    with log_path.open("w", encoding="utf-8") as log:
        process = subprocess.Popen(
            [sys.executable, "-m", "streamlit", "run", "app.py",
             "--server.address", "127.0.0.1", "--server.port", str(port),
             "--server.headless", "true"],
            cwd=str(folder), stdout=log, stderr=subprocess.STDOUT,
            creationflags=getattr(subprocess, "CREATE_NO_WINDOW", 0),
        )
    url = f"http://127.0.0.1:{port}"
    ready = False
    for _ in range(60):
        if process.poll() is not None:
            raise RuntimeError(log_path.read_text(encoding="utf-8", errors="replace"))
        try:
            with urllib.request.urlopen(url + "/_stcore/health", timeout=1) as response:
                ready = response.status == 200
            if ready:
                break
        except OSError:
            pass
        time.sleep(.25)
    if not ready:
        raise RuntimeError("服务尚未就绪，请查看Streamlit启动日志.txt")
display(Markdown(f"[点击打开动态看板]({url})"))
print("看板地址：", url)
print("演示期间保持Jupyter内核运行。")
```
